# AI Coffee Leaf Disease Detection — Abole

**Target accuracy: 92-95%** using:
- EfficientNet-B0 transfer learning (two-phase: freeze then full fine-tune)
- MixUp + CutMix augmentation
- Label smoothing
- Cosine annealing with warmup
- Test-Time Augmentation (TTA)
- Weighted loss for class imbalance

## 2. Load Dataset & EDA

```python
dataset = load_dataset('Project-AgML/arabica_coffee_leaf_disease_classification')
ds = dataset['train']
CLASS_NAMES = ds.features['label'].names
NUM_CLASSES = len(CLASS_NAMES)
print(f'Classes: {CLASS_NAMES}')
print(f'Total: {len(ds):,} images')

class_counts = {CLASS_NAMES[i]: sum(1 for ex in ds if ex['label'] == i) for i in range(NUM_CLASSES)}
print('\nClass distribution:')
for name, count in class_counts.items():
    print(f'  {name}: {count:,} ({count/len(ds)*100:.1f}%)')

fig, ax = plt.subplots(figsize=(9, 4))
ax.bar(class_counts.keys(), class_counts.values(), color=['#c0392b','#27ae60','#e67e22','#8e44ad','#2c3e50'])
ax.set_title('Class Distribution'); ax.set_ylabel('Images')
plt.tight_layout(); plt.savefig(SAVE_DIR+'class_dist.png', dpi=150); plt.show()
```

## 3. Data Leakage Prevention (Perceptual Hashing)

```python
print('Computing perceptual hashes (this takes ~5-10 mins)...')
hashes, labels_list = [], []
for i, ex in enumerate(ds):
    img = ex['image'].convert('RGB').resize((128, 128))
    hashes.append(imagehash.phash(img, hash_size=16))
    labels_list.append(ex['label'])
    if (i+1) % 5000 == 0: print(f'  {i+1}/{len(ds)}')

THRESHOLD = 10
visited, clusters = set(), []
for i in range(len(hashes)):
    if i in visited: continue
    cluster = {i}; visited.add(i)
    for j in range(i+1, len(hashes)):
        if j not in visited and hashes[i]-hashes[j] <= THRESHOLD:
            cluster.add(j); visited.add(j)
    clusters.append(cluster)
print(f'Found {len(clusters)} unique image clusters')

# Stratified 70/15/15 split at cluster level
cluster_labels = [Counter(labels_list[i] for i in c).most_common(1)[0][0] for c in clusters]
class_cluster_map = defaultdict(list)
for ci, lbl in enumerate(cluster_labels): class_cluster_map[lbl].append(ci)

train_c, val_c, test_c = set(), set(), set()
for lbl, idxs in class_cluster_map.items():
    random.shuffle(idxs)
    n = len(idxs)
    train_c.update(idxs[:int(n*0.70)])
    val_c.update(idxs[int(n*0.70):int(n*0.85)])
    test_c.update(idxs[int(n*0.85):])

train_idx, val_idx, test_idx = set(), set(), set()
for ci, cluster in enumerate(clusters):
    if ci in train_c: train_idx.update(cluster)
    elif ci in val_c: val_idx.update(cluster)
    else: test_idx.update(cluster)

print(f'Train: {len(train_idx):,} | Val: {len(val_idx):,} | Test: {len(test_idx):,}')
```

## 4. Augmentation Pipeline (Strong augmentation for higher accuracy)

```python
MEAN = [0.485, 0.456, 0.406]; STD = [0.229, 0.224, 0.225]

# STRONG augmentation for training
train_transform = T.Compose([
    T.Resize((256, 256)),
    T.RandomCrop(224),
    T.RandomHorizontalFlip(),
    T.RandomVerticalFlip(),
    T.RandomRotation(20),
    T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.05),
    T.RandomAffine(degrees=0, translate=(0.1, 0.1), scale=(0.85, 1.15)),
    T.RandomGrayscale(p=0.05),
    T.ToTensor(),
    T.Normalize(MEAN, STD),
    T.RandomErasing(p=0.2, scale=(0.02, 0.1)),
])

eval_transform = T.Compose([
    T.Resize((224, 224)),
    T.ToTensor(),
    T.Normalize(MEAN, STD),
])

# TTA transforms (multiple crops + flips at inference)
tta_transforms = [
    T.Compose([T.Resize((224,224)), T.ToTensor(), T.Normalize(MEAN,STD)]),
    T.Compose([T.Resize((224,224)), T.RandomHorizontalFlip(p=1.0), T.ToTensor(), T.Normalize(MEAN,STD)]),
    T.Compose([T.Resize((256,256)), T.CenterCrop(224), T.ToTensor(), T.Normalize(MEAN,STD)]),
    T.Compose([T.Resize((256,256)), T.CenterCrop(224), T.RandomHorizontalFlip(p=1.0), T.ToTensor(), T.Normalize(MEAN,STD)]),
    T.Compose([T.Resize((224,224)), T.RandomVerticalFlip(p=1.0), T.ToTensor(), T.Normalize(MEAN,STD)]),
]

class CoffeeDataset(Dataset):
    def __init__(self, hf_ds, indices, transform, oversample=None, factor=2):
        self.ds = hf_ds
        self.transform = transform
        idxs = list(indices)
        if oversample:
            extra = [i for i in idxs if hf_ds[i]['label'] in oversample]
            idxs += extra * (factor - 1)
        self.indices = idxs
    def __len__(self): return len(self.indices)
    def __getitem__(self, i):
        ex = self.ds[self.indices[i]]
        return self.transform(ex['image'].convert('RGB')), ex['label']

BATCH = 32
train_ds = CoffeeDataset(ds, train_idx, train_transform, oversample=[0,4], factor=2)
val_ds   = CoffeeDataset(ds, val_idx,   eval_transform)
test_ds  = CoffeeDataset(ds, test_idx,  eval_transform)

train_dl = DataLoader(train_ds, batch_size=BATCH, shuffle=True,  num_workers=2, pin_memory=True)
val_dl   = DataLoader(val_ds,   batch_size=BATCH, shuffle=False, num_workers=2, pin_memory=True)
test_dl  = DataLoader(test_ds,  batch_size=BATCH, shuffle=False, num_workers=2, pin_memory=True)
print(f'Train: {len(train_ds):,} | Val: {len(val_ds):,} | Test: {len(test_ds):,}')
```

## 5. MixUp & CutMix (boosts accuracy 2-4%)

```python
def mixup_data(x, y, alpha=0.4):
    lam = np.random.beta(alpha, alpha) if alpha > 0 else 1.0
    idx = torch.randperm(x.size(0), device=x.device)
    mixed_x = lam * x + (1 - lam) * x[idx]
    return mixed_x, y, y[idx], lam

def cutmix_data(x, y, alpha=1.0):
    lam = np.random.beta(alpha, alpha)
    idx = torch.randperm(x.size(0), device=x.device)
    _, _, H, W = x.shape
    cut_rat = np.sqrt(1 - lam)
    cut_w, cut_h = int(W * cut_rat), int(H * cut_rat)
    cx, cy = np.random.randint(W), np.random.randint(H)
    x1, x2 = max(cx - cut_w//2, 0), min(cx + cut_w//2, W)
    y1, y2 = max(cy - cut_h//2, 0), min(cy + cut_h//2, H)
    mixed_x = x.clone()
    mixed_x[:, :, y1:y2, x1:x2] = x[idx, :, y1:y2, x1:x2]
    lam = 1 - (x2-x1)*(y2-y1)/(W*H)
    return mixed_x, y, y[idx], lam

def mixup_criterion(criterion, pred, ya, yb, lam):
    return lam * criterion(pred, ya) + (1 - lam) * criterion(pred, yb)

print('MixUp and CutMix functions ready')
```

## 6. Model + Class Weights + Label Smoothing

```python
# EfficientNet-B0 with pretrained ImageNet weights
model = timm.create_model('efficientnet_b0', pretrained=True, num_classes=NUM_CLASSES)
model = model.to(DEVICE)

# Class weights (inverse frequency)
counts = torch.tensor([class_counts[CLASS_NAMES[i]] for i in range(NUM_CLASSES)], dtype=torch.float32)
weights = (1.0 / counts); weights = weights / weights.sum() * NUM_CLASSES
weights = weights.to(DEVICE)
print('Class weights:')
for i, n in enumerate(CLASS_NAMES): print(f'  {n}: {weights[i].item():.4f}')

# Label smoothing reduces overconfidence and improves generalization
criterion = nn.CrossEntropyLoss(weight=weights, label_smoothing=0.1)

total = sum(p.numel() for p in model.parameters())
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f'\nParameters: {total:,} total | {trainable:,} trainable')
```

## 7. Two-Phase Training

**Phase 1 (5 epochs):** Freeze backbone, train only classifier head — fast warm-up.
**Phase 2 (25 epochs):** Unfreeze all layers, fine-tune with low LR + cosine schedule.

```python
def freeze_backbone(model):
    for name, p in model.named_parameters():
        p.requires_grad = 'classifier' in name or 'head' in name or 'bn2' in name
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f'Phase 1 — trainable: {trainable:,} params (classifier only)')

def unfreeze_all(model):
    for p in model.parameters(): p.requires_grad = True
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f'Phase 2 — trainable: {trainable:,} params (all layers)')

def train_epoch(model, loader, criterion, optimizer, scaler, use_mixup=True):
    model.train()
    loss_sum, correct, total = 0.0, 0, 0
    for imgs, lbls in loader:
        imgs, lbls = imgs.to(DEVICE), lbls.to(DEVICE)
        # Randomly choose MixUp or CutMix
        if use_mixup and random.random() > 0.5:
            if random.random() > 0.5:
                imgs, ya, yb, lam = mixup_data(imgs, lbls)
            else:
                imgs, ya, yb, lam = cutmix_data(imgs, lbls)
            optimizer.zero_grad()
            with torch.cuda.amp.autocast():
                out = model(imgs)
                loss = mixup_criterion(criterion, out, ya, yb, lam)
        else:
            optimizer.zero_grad()
            with torch.cuda.amp.autocast():
                out = model(imgs)
                loss = criterion(out, lbls)
        scaler.scale(loss).backward()
        scaler.step(optimizer); scaler.update()
        loss_sum += loss.item()
        correct += out.argmax(1).eq(lbls).sum().item()
        total += lbls.size(0)
    return loss_sum/len(loader), correct/total

def eval_epoch(model, loader, criterion):
    model.eval()
    loss_sum, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for imgs, lbls in loader:
            imgs, lbls = imgs.to(DEVICE), lbls.to(DEVICE)
            with torch.cuda.amp.autocast():
                out = model(imgs); loss = criterion(out, lbls)
            loss_sum += loss.item()
            correct += out.argmax(1).eq(lbls).sum().item()
            total += lbls.size(0)
    return loss_sum/len(loader), correct/total

scaler = torch.cuda.amp.GradScaler()
history = {'train_loss':[],'val_loss':[],'train_acc':[],'val_acc':[]}
best_val_loss = float('inf')
patience_counter = 0
PATIENCE = 8

# ── PHASE 1: Warm-up (5 epochs, classifier only, high LR) ──
print('\n=== PHASE 1: Classifier warm-up ===')
freeze_backbone(model)
optimizer = optim.AdamW(filter(lambda p: p.requires_grad, model.parameters()), lr=3e-3, weight_decay=1e-4)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=5)

for epoch in range(5):
    tl, ta = train_epoch(model, train_dl, criterion, optimizer, scaler, use_mixup=False)
    vl, va = eval_epoch(model, val_dl, criterion)
    scheduler.step()
    history['train_loss'].append(tl); history['val_loss'].append(vl)
    history['train_acc'].append(ta); history['val_acc'].append(va)
    print(f'[P1] Epoch {epoch+1}/5 | Train Loss:{tl:.4f} Acc:{ta:.4f} | Val Loss:{vl:.4f} Acc:{va:.4f}')

# ── PHASE 2: Full fine-tuning (25 epochs, all layers, low LR + MixUp) ──
print('\n=== PHASE 2: Full fine-tuning ===')
unfreeze_all(model)
optimizer = optim.AdamW(model.parameters(), lr=5e-5, weight_decay=1e-4)
scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2)

for epoch in range(25):
    tl, ta = train_epoch(model, train_dl, criterion, optimizer, scaler, use_mixup=True)
    vl, va = eval_epoch(model, val_dl, criterion)
    scheduler.step()
    history['train_loss'].append(tl); history['val_loss'].append(vl)
    history['train_acc'].append(ta); history['val_acc'].append(va)
    print(f'[P2] Epoch {epoch+1}/25 | Train Loss:{tl:.4f} Acc:{ta:.4f} | Val Loss:{vl:.4f} Acc:{va:.4f} | LR:{scheduler.get_last_lr()[0]:.6f}')
    if vl < best_val_loss:
        best_val_loss = vl; patience_counter = 0
        torch.save({'epoch': epoch+1, 'model_state_dict': model.state_dict(), 'val_loss': vl, 'val_acc': va},
                   SAVE_DIR+'best_model.pth')
        print(f'  -> Best model saved (val_loss={vl:.4f}, val_acc={va:.4f})')
    else:
        patience_counter += 1
    if patience_counter >= PATIENCE:
        print(f'Early stopping at epoch {epoch+1}'); break

print('\nTraining complete!')
```

```python
# Plot training curves
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(history['train_loss'], label='Train', color='#2d5016')
ax1.plot(history['val_loss'], label='Val', color='#d4a017')
ax1.axvline(x=4, color='gray', linestyle='--', label='Phase 2 start')
ax1.set_title('Loss'); ax1.legend()
ax2.plot(history['train_acc'], label='Train', color='#2d5016')
ax2.plot(history['val_acc'], label='Val', color='#d4a017')
ax2.axvline(x=4, color='gray', linestyle='--', label='Phase 2 start')
ax2.set_title('Accuracy'); ax2.legend()
plt.tight_layout()
plt.savefig(SAVE_DIR+'training_curves.png', dpi=150); plt.show()
```

## 8. Evaluation with Test-Time Augmentation (TTA +1-2% accuracy)

```python
# Load best model
ckpt = torch.load(SAVE_DIR+'best_model.pth', map_location=DEVICE)
model.load_state_dict(ckpt['model_state_dict'])
model.eval()
print(f'Loaded best model — epoch {ckpt["epoch"]}, val_loss={ckpt["val_loss"]:.4f}, val_acc={ckpt["val_acc"]:.4f}')

# ── Standard evaluation ──
all_preds, all_labels = [], []
with torch.no_grad():
    for imgs, lbls in test_dl:
        out = model(imgs.to(DEVICE))
        all_preds.extend(out.argmax(1).cpu().numpy())
        all_labels.extend(lbls.numpy())
all_preds, all_labels = np.array(all_preds), np.array(all_labels)
std_acc = (all_preds == all_labels).mean()
print(f'\nStandard accuracy: {std_acc:.4f} ({std_acc*100:.2f}%)')

# ── TTA evaluation ──
print('\nRunning Test-Time Augmentation (5 passes)...')
test_raw = list(test_idx)

all_tta_probs = np.zeros((len(test_raw), NUM_CLASSES))

for t_idx, tfm in enumerate(tta_transforms):
    print(f'  TTA pass {t_idx+1}/{len(tta_transforms)}')
    tta_ds = CoffeeDataset(ds, test_raw, tfm)
    tta_dl = DataLoader(tta_ds, batch_size=BATCH, shuffle=False, num_workers=2)
    probs_list = []
    with torch.no_grad():
        for imgs, _ in tta_dl:
            out = model(imgs.to(DEVICE))
            probs_list.extend(F.softmax(out, dim=1).cpu().numpy())
    all_tta_probs += np.array(probs_list)

tta_preds = all_tta_probs.argmax(axis=1)
tta_labels = [ds[i]['label'] for i in test_raw]
tta_acc = (tta_preds == np.array(tta_labels)).mean()
print(f'\nTTA accuracy: {tta_acc:.4f} ({tta_acc*100:.2f}%)')
print(f'TTA improvement: +{(tta_acc - std_acc)*100:.2f}%')
```

```python
# Full classification report
report = classification_report(all_labels, all_preds, target_names=CLASS_NAMES, digits=4)
print('Classification Report:'); print('='*70); print(report)

# Confusion matrix
cm = confusion_matrix(all_labels, all_preds)
fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, ax=ax)
ax.set_xlabel('Predicted'); ax.set_ylabel('Actual')
ax.set_title(f'Confusion Matrix (Test Accuracy: {std_acc*100:.2f}%)')
plt.tight_layout(); plt.savefig(SAVE_DIR+'confusion_matrix.png', dpi=150); plt.show()

print('\nPer-class accuracy:')
for i, name in enumerate(CLASS_NAMES):
    mask = all_labels == i
    acc = (all_preds[mask] == all_labels[mask]).mean()
    print(f'  {name}: {acc*100:.2f}%')
```
